**Tujuan:** Menganalisis titik-titik koordinat fasilitas atau titik potensi pasar (store location analysis) menggunakan Python di Google Colab dengan pustaka geospasial (geopandas / folium), merefleksikan bagaimana data spasial mendukung data-driven decision making.

```python
# 1. Install & Import Library Pendukung Spasial
!pip install folium pandas numpy

import folium
import pandas as pd
import numpy as np

# 2. Simulasi Data Koordinat Lokasi Fasilitas / Toko di Kota Bengkulu
np.random.seed(42)
n_points = 15

# Titik pusat simulasi (Kota Bengkulu)
base_lat, base_lon = -3.7928, 102.2608

df_gis = pd.DataFrame({
    'Nama_Lokasi': [f'Cabang / Titik_{i+1}' for i in range(n_points)],
    'Latitude': base_lat + np.random.uniform(-0.05, 0.05, n_points),
    'Longitude': base_lon + np.random.uniform(-0.05, 0.05, n_points),
    'Jumlah_Transaksi_Harian': np.random.randint(100, 850, n_points)
})

print("Preview Data Spasial:")
display(df_gis.head())

# 3. Visualisasi Peta Interaktif menggunakan Folium (Web Map / Cloud GIS Preview)
m = folium.Map(location=[base_lat, base_lon], zoom_start=13)

for idx, row in df_gis.iterrows():
    # Ukuran lingkaran marker merepresentasikan volume transaksi harian
    radius_size = row['Jumlah_Transaksi_Harian'] / 35
    folium.CircleMarker(
        location=[row['Latitude'], row['Longitude']],
        radius=radius_size,
        popup=f"<b>{row['Nama_Lokasi']}</b><br>Transaksi: {row['Jumlah_Transaksi_Harian']}",
        color='blue',
        fill=True,
        fill_color='cyan',
        fill_opacity=0.6
    ).add_to(m)

# Tampilkan Peta Interaktif di Colab
m
```
