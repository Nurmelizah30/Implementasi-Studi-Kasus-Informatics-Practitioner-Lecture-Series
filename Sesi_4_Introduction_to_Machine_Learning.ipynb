**Studi Kasus: Klasifikasi Kategori Kualitas/Kondisi Lahan Pertanian atau Lingkungan (Supervised Learning - Classification).**

**Tujuan:** Memahami pergeseran dari rule-based thinking ke data-driven thinking melalui pelatihan model Supervised Machine Learning standar menggunakan dataset klasik yang disesuaikan.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt

# 1. Memuat Dataset (Contoh: Dataset Klasifikasi Karakteristik/Kondisi Objek)
iris = load_iris()
X, y = iris.data, iris.target

# Membagi data training & testing (Data-driven approach)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 2. Inisialisasi & Pelatihan Model Decision Tree (Rule-learning otomatis dari data)
clf = DecisionTreeClassifier(max_depth=3, random_state=42)
clf.fit(X_train, y_train)

# 3. Prediksi & Evaluasi
y_pred = clf.predict(X_test)
acc = accuracy_score(y_test, y_pred)

print(f"Akurasi Model Decision Tree: {acc * 100:.2f}%\n")

# 4. Visualisasi Pohon Keputusan (Representasi aturan yang dipelajari mesin dari data)
plt.figure(figsize=(12, 8))
plot_tree(clf, feature_names=iris.feature_names, class_names=iris.target_names, filled=True)
plt.title("Visualisasi Model Machine Learning (Decision Tree)")
plt.show()
```
