**Studi Kasus: Simulasi Komputasi Paralel untuk Pemodelan Evakuasi / Dinamika Kerumunan (Agent-Based Crowd / Evacuation Simulation).**

**Tujuan:** Mensimulasikan pemrosesan beban berat secara paralel menggunakan Multiprocessing di Python (menggambarkan konsep pemecahan tugas besar yang biasa dijalankan di kluster HPC seperti Lanta/Fugaku).

```python
import time
from multiprocessing import Pool, cpu_count
import numpy as np

print(f"Jumlah CPU Core yang tersedia di Colab: {cpu_count()}")

# Fungsi berat (simulasi komputasi intensif / pemodelan evakuasi grid seluler)
def simulate_evacuation_grid(grid_seed):
    np.random.seed(grid_seed)
    # Simulasi perhitungan matriks besar / iterasi model kerumunan
    matrix = np.random.rand(1500, 1500)
    result = np.linalg.inv(matrix @ matrix.T)
    return np.sum(result)

# Dataset tugas simulasi (misal 8 skenario wilayah evakuasi)
tasks = [101, 102, 103, 104, 105, 106, 107, 108]

# 1. Eksekusi Sekuensial (Tanpa Paralelisasi / Single Node biasa)
start_time = time.time()
seq_results = [simulate_evacuation_grid(t) for t in tasks]
seq_duration = time.time() - start_time
print(f"Waktu Eksekusi Sekuensial: {seq_duration:.2f} detik")

# 2. Eksekusi Paralel (Mengoptimalkan Multi-core ala HPC)
start_time = time.time()
with Pool(processes=4) as pool:
    par_results = pool.map(simulate_evacuation_grid, tasks)
par_duration = time.time() - start_time
print(f"Waktu Eksekusi Paralel (HPC Simulation Style): {par_duration:.2f} detik")
print(f"Speedup Ratio: {seq_duration / par_duration:.2f}x lebih cepat!")
```
